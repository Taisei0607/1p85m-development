```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
lambda115 = 3.00/1.15
lambda230 = 3.00/2.30
lambda345 = 3.00/3.45

omega_sub = 76
r_sub = 1105.97

l_sub_M1 = 1197.83 # 1
l_sub_M2 = 1400 # 2
l_M1_M2 = 202.17 # 3
l_M2_M3 = 785 # 4
l_M3_M4 = 180 # 5
l_M4_fil = 230.388 # 6
l_fil_M5 = 110 # 7
l_fil_M6 = 60 # 8
l_M4_M5 = l_M4_fil + l_fil_M5 # 9
l_M4_M6 = l_M4_fil + l_fil_M6 # 10
l_M5_horn = 60 # 11
l_M6_horn = 30 # 12
print(l_M4_M5, l_M4_M6)

f_M2 = 290
f_M3 = 100
f_M4 = 60
f_M5 = 80
f_M6 = 34
```

```python
z_sub_omega01_115 = r_sub/(1+((lambda115*r_sub)/(np.pi*(omega_sub**2)))**2)
z_sub_omega01_230 = r_sub/(1+((lambda230*r_sub)/(np.pi*(omega_sub**2)))**2)
z_sub_omega01_345 = r_sub/(1+((lambda345*r_sub)/(np.pi*(omega_sub**2)))**2)

z_omega01_115_M2 = l_sub_M2-z_sub_omega01_115
z_omega01_230_M2 = l_sub_M2-z_sub_omega01_230
z_omega01_345_M2 = l_sub_M2-z_sub_omega01_345

omega01_115 = omega_sub/np.sqrt(1+((np.pi*(omega_sub**2))/(lambda115*r_sub))**2)
omega01_230 = omega_sub/np.sqrt(1+((np.pi*(omega_sub**2))/(lambda230*r_sub))**2)
omega01_345 = omega_sub/np.sqrt(1+((np.pi*(omega_sub**2))/(lambda345*r_sub))**2)

omega_M2_115 = omega01_115*np.sqrt(1+((lambda115*z_omega01_115_M2)/(np.pi*(omega01_115**2)))**2)
omega_M2_230 = omega01_230*np.sqrt(1+((lambda230*z_omega01_230_M2)/(np.pi*(omega01_230**2)))**2)
omega_M2_345 = omega01_345*np.sqrt(1+((lambda345*z_omega01_345_M2)/(np.pi*(omega01_345**2)))**2)

r_M2_bef_115 = z_omega01_115_M2*(1+((np.pi*(omega01_115**2))/(lambda115*z_omega01_115_M2))**2)
r_M2_bef_230 = z_omega01_230_M2*(1+((np.pi*(omega01_230**2))/(lambda230*z_omega01_230_M2))**2)
r_M2_bef_345 = z_omega01_345_M2*(1+((np.pi*(omega01_345**2))/(lambda345*z_omega01_345_M2))**2)

r_M2_aft_115 = (f_M2*r_M2_bef_115)/(r_M2_bef_115-f_M2)
r_M2_aft_230 = (f_M2*r_M2_bef_230)/(r_M2_bef_230-f_M2)
r_M2_aft_345 = (f_M2*r_M2_bef_345)/(r_M2_bef_345-f_M2)

z_M2_omega02_115 = r_M2_aft_115/(1+((lambda115*r_M2_aft_115)/(np.pi*(omega_M2_115**2)))**2)
z_M2_omega02_230 = r_M2_aft_230/(1+((lambda230*r_M2_aft_230)/(np.pi*(omega_M2_230**2)))**2)
z_M2_omega02_345 = r_M2_aft_345/(1+((lambda345*r_M2_aft_345)/(np.pi*(omega_M2_345**2)))**2)

z_omega02_115_M3 = l_M2_M3-z_M2_omega02_115
z_omega02_230_M3 = l_M2_M3-z_M2_omega02_230
z_omega02_345_M3 = l_M2_M3-z_M2_omega02_345

omega02_115 = omega_M2_115/np.sqrt(1+((np.pi*(omega_M2_115**2))/(lambda115*r_M2_aft_115))**2)
omega02_230 = omega_M2_230/np.sqrt(1+((np.pi*(omega_M2_230**2))/(lambda230*r_M2_aft_230))**2)
omega02_345 = omega_M2_345/np.sqrt(1+((np.pi*(omega_M2_345**2))/(lambda345*r_M2_aft_345))**2)

omega_M3_115 = omega02_115*np.sqrt(1+((lambda115*z_omega02_115_M3)/(np.pi*(omega02_115**2)))**2)
omega_M3_230 = omega02_230*np.sqrt(1+((lambda230*z_omega02_230_M3)/(np.pi*(omega02_230**2)))**2)
omega_M3_345 = omega02_345*np.sqrt(1+((lambda345*z_omega02_345_M3)/(np.pi*(omega02_345**2)))**2)

r_M3_bef_115 = z_omega02_115_M3*(1+((np.pi*(omega02_115**2))/(lambda115*z_omega02_115_M3))**2)
r_M3_bef_230 = z_omega02_230_M3*(1+((np.pi*(omega02_230**2))/(lambda230*z_omega02_230_M3))**2)
r_M3_bef_345 = z_omega02_345_M3*(1+((np.pi*(omega02_345**2))/(lambda345*z_omega02_345_M3))**2)

r_M3_aft_115 = (f_M3*r_M3_bef_115)/(r_M3_bef_115-f_M3)
r_M3_aft_230 = (f_M3*r_M3_bef_230)/(r_M3_bef_230-f_M3)
r_M3_aft_345 = (f_M3*r_M3_bef_345)/(r_M3_bef_345-f_M3)

z_M3_omega03_115 = r_M3_aft_115/(1+((lambda115*r_M3_aft_115)/(np.pi*(omega_M3_115**2)))**2)
z_M3_omega03_230 = r_M3_aft_230/(1+((lambda230*r_M3_aft_230)/(np.pi*(omega_M3_230**2)))**2)
z_M3_omega03_345 = r_M3_aft_345/(1+((lambda345*r_M3_aft_345)/(np.pi*(omega_M3_345**2)))**2)

z_omega03_115_M4 = l_M3_M4-z_M3_omega03_115
z_omega03_230_M4 = l_M3_M4-z_M3_omega03_230
z_omega03_345_M4 = l_M3_M4-z_M3_omega03_345

omega03_115 = omega_M3_115/np.sqrt(1+((np.pi*(omega_M3_115**2))/(lambda115*r_M3_aft_115))**2)
omega03_230 = omega_M3_230/np.sqrt(1+((np.pi*(omega_M3_230**2))/(lambda230*r_M3_aft_230))**2)
omega03_345 = omega_M3_345/np.sqrt(1+((np.pi*(omega_M3_345**2))/(lambda345*r_M3_aft_345))**2)

omega_M4_115 = omega03_115*np.sqrt(1+((lambda115*z_omega03_115_M4)/(np.pi*(omega03_115**2)))**2)
omega_M4_230 = omega03_230*np.sqrt(1+((lambda230*z_omega03_230_M4)/(np.pi*(omega03_230**2)))**2)
omega_M4_345 = omega03_345*np.sqrt(1+((lambda345*z_omega03_345_M4)/(np.pi*(omega03_345**2)))**2)

r_M4_bef_115 = z_omega03_115_M4*(1+((np.pi*(omega03_115**2))/(lambda115*z_omega03_115_M4))**2)
r_M4_bef_230 = z_omega03_230_M4*(1+((np.pi*(omega03_230**2))/(lambda230*z_omega03_230_M4))**2)
r_M4_bef_345 = z_omega03_345_M4*(1+((np.pi*(omega03_345**2))/(lambda345*z_omega03_345_M4))**2)

r_M4_aft_115 = (f_M4*r_M4_bef_115)/(r_M4_bef_115-f_M4)
r_M4_aft_230 = (f_M4*r_M4_bef_230)/(r_M4_bef_230-f_M4)
r_M4_aft_345 = (f_M4*r_M4_bef_345)/(r_M4_bef_345-f_M4)

z_M4_omega04_115 = r_M4_aft_115/(1+((lambda115*r_M4_aft_115)/(np.pi*(omega_M4_115**2)))**2)
z_omega04_115_M5 = (l_M4_fil+l_fil_M5)-z_M4_omega04_115
z_M4_omega04_230 = r_M4_aft_230/(1+((lambda230*r_M4_aft_230)/(np.pi*(omega_M4_230**2)))**2)
z_omega04_230_M6 = (l_M4_fil+l_fil_M6)-z_M4_omega04_230
z_M4_omega04_345 = r_M4_aft_345/(1+((lambda345*r_M4_aft_345)/(np.pi*(omega_M4_345**2)))**2)
z_omega04_345_M6 = (l_M4_fil+l_fil_M6)-z_M4_omega04_345

omega04_115 = omega_M4_115/np.sqrt(1+((np.pi*(omega_M4_115**2))/(lambda115*r_M4_aft_115))**2)
omega04_230 = omega_M4_230/np.sqrt(1+((np.pi*(omega_M4_230**2))/(lambda230*r_M4_aft_230))**2)
omega04_345 = omega_M4_345/np.sqrt(1+((np.pi*(omega_M4_345**2))/(lambda345*r_M4_aft_345))**2)

omega_M5_115 = omega04_115*np.sqrt(1+((lambda115*z_omega04_115_M5)/(np.pi*(omega04_115**2)))**2)
omega_M6_230 = omega04_230*np.sqrt(1+((lambda230*z_omega04_230_M6)/(np.pi*(omega04_230**2)))**2)
omega_M6_345 = omega04_345*np.sqrt(1+((lambda345*z_omega04_345_M6)/(np.pi*(omega04_345**2)))**2)

r_M5_bef_115 = z_omega04_115_M5*(1+((np.pi*(omega04_115**2))/(lambda115*z_omega04_115_M5))**2)
r_M6_bef_230 = z_omega04_230_M6*(1+((np.pi*(omega04_230**2))/(lambda230*z_omega04_230_M6))**2)
r_M6_bef_345 = z_omega04_345_M6*(1+((np.pi*(omega04_345**2))/(lambda345*z_omega04_345_M6))**2)

r_M5_aft_115 = (f_M5*r_M5_bef_115)/(r_M5_bef_115-f_M5)
r_M6_aft_230 = (f_M6*r_M6_bef_230)/(r_M6_bef_230-f_M6)
r_M6_aft_345 = (f_M6*r_M6_bef_345)/(r_M6_bef_345-f_M6)

z_M5_omega05_115 = r_M5_aft_115/(1+((lambda115*r_M5_aft_115)/(np.pi*(omega_M5_115**2)))**2)
z_omega05_115_horn = l_M5_horn-z_M5_omega05_115
z_M6_omega05_230 = r_M6_aft_230/(1+((lambda230*r_M6_aft_230)/(np.pi*(omega_M6_230**2)))**2)
z_omega05_230_horn = l_M6_horn-z_M6_omega05_230
z_M6_omega05_345 = r_M6_aft_345/(1+((lambda345*r_M6_aft_345)/(np.pi*(omega_M6_345**2)))**2)
z_omega05_345_horn = l_M6_horn-z_M6_omega05_345

omega05_115 = omega_M5_115/np.sqrt(1+((np.pi*(omega_M5_115**2))/(lambda115*r_M5_aft_115))**2)
omega05_230 = omega_M6_230/np.sqrt(1+((np.pi*(omega_M6_230**2))/(lambda230*r_M6_aft_230))**2)
omega05_345 = omega_M6_345/np.sqrt(1+((np.pi*(omega_M6_345**2))/(lambda345*r_M6_aft_345))**2)

omega_horn_115 = omega05_115*np.sqrt(1+((lambda115*z_omega05_115_horn)/(np.pi*(omega05_115**2)))**2)
omega_horn_230 = omega05_230*np.sqrt(1+((lambda230*z_omega05_230_horn)/(np.pi*(omega05_230**2)))**2)
omega_horn_345 = omega05_345*np.sqrt(1+((lambda345*z_omega05_345_horn)/(np.pi*(omega05_345**2)))**2)

r_horn_bef_115 = abs(z_omega05_115_horn)*(1+((np.pi*(omega05_115**2))/(lambda115*abs(z_omega05_115_horn)))**2)
r_horn_bef_230 = abs(z_omega05_230_horn)*(1+((np.pi*(omega05_230**2))/(lambda230*abs(z_omega05_230_horn)))**2)
r_horn_bef_345 = abs(z_omega05_345_horn)*(1+((np.pi*(omega05_345**2))/(lambda345*abs(z_omega05_345_horn)))**2)

r_115 = omega_horn_115/0.76
l_115 = r_horn_bef_115
theta_115 = np.arcsin(r_115/l_115)*(180/np.pi)
r_230 = omega_horn_230/0.76
l_230 = r_horn_bef_230
theta_230 = np.arcsin(r_230/l_230)*(180/np.pi)
r_345 = omega_horn_345/0.76
l_345 = r_horn_bef_345
theta_345 = np.arcsin(r_345/l_345)*(180/np.pi)

x1_115 = np.linspace(0, z_sub_omega01_115+z_omega01_115_M2)
x2_115 = np.linspace(z_sub_omega01_115+z_omega01_115_M2, z_sub_omega01_115+z_omega01_115_M2+z_M2_omega02_115+
    z_omega02_115_M3)
x3_115 = np.linspace(z_sub_omega01_115+z_omega01_115_M2+z_M2_omega02_115+z_omega02_115_M3, z_sub_omega01_115+
    z_omega01_115_M2+z_M2_omega02_115+z_omega02_115_M3+z_M3_omega03_115+z_omega03_115_M4)
x4_115 = np.linspace(z_sub_omega01_115+z_omega01_115_M2+z_M2_omega02_115+z_omega02_115_M3+z_M3_omega03_115+
    z_omega03_115_M4, z_sub_omega01_115+z_omega01_115_M2+z_M2_omega02_115+z_omega02_115_M3+z_M3_omega03_115+
    z_omega03_115_M4+z_M4_omega04_115+z_omega04_115_M5)
x5_115 = np.linspace(z_sub_omega01_115+z_omega01_115_M2+z_M2_omega02_115+z_omega02_115_M3+z_M3_omega03_115+
    z_omega03_115_M4+z_M4_omega04_115+z_omega04_115_M5, z_sub_omega01_115+z_omega01_115_M2+z_M2_omega02_115+
    z_omega02_115_M3+z_M3_omega03_115+z_omega03_115_M4+z_M4_omega04_115+z_omega04_115_M5+z_M5_omega05_115+
    z_omega05_115_horn)

x1_230 = np.linspace(0, z_sub_omega01_230+z_omega01_230_M2)
x2_230 = np.linspace(z_sub_omega01_230+z_omega01_230_M2, z_sub_omega01_230+z_omega01_230_M2+z_M2_omega02_230+
    z_omega02_230_M3)
x3_230 = np.linspace(z_sub_omega01_230+z_omega01_230_M2+z_M2_omega02_230+z_omega02_230_M3, z_sub_omega01_230+
    z_omega01_230_M2+z_M2_omega02_230+z_omega02_230_M3+z_M3_omega03_230+z_omega03_230_M4)
x4_230 = np.linspace(z_sub_omega01_230+z_omega01_230_M2+z_M2_omega02_230+z_omega02_230_M3+z_M3_omega03_230+
    z_omega03_230_M4, z_sub_omega01_230+z_omega01_230_M2+z_M2_omega02_230+z_omega02_230_M3+z_M3_omega03_230+
    z_omega03_230_M4+z_M4_omega04_230+z_omega04_230_M6)
x5_230 = np.linspace(z_sub_omega01_230+z_omega01_230_M2+z_M2_omega02_230+z_omega02_230_M3+z_M3_omega03_230+
    z_omega03_230_M4+z_M4_omega04_230+z_omega04_230_M6, z_sub_omega01_230+z_omega01_230_M2+z_M2_omega02_230+
    z_omega02_230_M3+z_M3_omega03_230+z_omega03_230_M4+z_M4_omega04_230+z_omega04_230_M6+z_M6_omega05_230+
    z_omega05_230_horn)

x1_345 = np.linspace(0, z_sub_omega01_345+z_omega01_345_M2)
x2_345 = np.linspace(z_sub_omega01_345+z_omega01_345_M2, z_sub_omega01_345+z_omega01_345_M2+z_M2_omega02_345+
    z_omega02_345_M3)
x3_345 = np.linspace(z_sub_omega01_345+z_omega01_345_M2+z_M2_omega02_345+z_omega02_345_M3, z_sub_omega01_345+
    z_omega01_345_M2+z_M2_omega02_345+z_omega02_345_M3+z_M3_omega03_345+z_omega03_345_M4)
x4_345 = np.linspace(z_sub_omega01_345+z_omega01_345_M2+z_M2_omega02_345+z_omega02_345_M3+z_M3_omega03_345+
    z_omega03_345_M4, z_sub_omega01_345+z_omega01_345_M2+z_M2_omega02_345+z_omega02_345_M3+z_M3_omega03_345+
    z_omega03_345_M4+z_M4_omega04_345+z_omega04_345_M6)
x5_345 = np.linspace(z_sub_omega01_345+z_omega01_345_M2+z_M2_omega02_345+z_omega02_345_M3+z_M3_omega03_345+
    z_omega03_345_M4+z_M4_omega04_345+z_omega04_345_M6, z_sub_omega01_345+z_omega01_345_M2+z_M2_omega02_345+
    z_omega02_345_M3+z_M3_omega03_345+z_omega03_345_M4+z_M4_omega04_345+z_omega04_345_M6+z_M6_omega05_345+
    z_omega05_345_horn)

y1_115 = omega01_115*np.sqrt(1+((lambda115*(x1_115-z_sub_omega01_115))/(np.pi*(omega01_115**2)))**2)
y2_115 = omega02_115*np.sqrt(1+((lambda115*(x2_115-z_sub_omega01_115-z_omega01_115_M2-z_M2_omega02_115))/
    (np.pi*(omega02_115**2)))**2)
y3_115 = omega03_115*np.sqrt(1+((lambda115*(x3_115-z_sub_omega01_115-z_omega01_115_M2-z_M2_omega02_115-
    z_omega02_115_M3-z_M3_omega03_115))/(np.pi*(omega03_115**2)))**2)
y4_115 = omega04_115*np.sqrt(1+((lambda115*(x4_115-z_sub_omega01_115-z_omega01_115_M2-z_M2_omega02_115-
    z_omega02_115_M3-z_M3_omega03_115-z_omega03_115_M4-z_M4_omega04_115))/(np.pi*(omega04_115**2)))**2)
y5_115 = omega05_115*np.sqrt(1+((lambda115*(x5_115-z_sub_omega01_115-z_omega01_115_M2-z_M2_omega02_115-
    z_omega02_115_M3-z_M3_omega03_115-z_omega03_115_M4-z_M4_omega04_115-z_omega04_115_M5-z_M5_omega05_115))/
    (np.pi*(omega05_115**2)))**2)

y1_230 = omega01_230*np.sqrt(1+((lambda230*(x1_230-z_sub_omega01_230))/(np.pi*(omega01_230**2)))**2)
y2_230 = omega02_230*np.sqrt(1+((lambda230*(x2_230-z_sub_omega01_230-z_omega01_230_M2-z_M2_omega02_230))/
    (np.pi*(omega02_230**2)))**2)
y3_230 = omega03_230*np.sqrt(1+((lambda230*(x3_230-z_sub_omega01_230-z_omega01_230_M2-z_M2_omega02_230-
    z_omega02_230_M3-z_M3_omega03_230))/(np.pi*(omega03_230**2)))**2)
y4_230 = omega04_230*np.sqrt(1+((lambda230*(x4_230-z_sub_omega01_230-z_omega01_230_M2-z_M2_omega02_230-
    z_omega02_230_M3-z_M3_omega03_230-z_omega03_230_M4-z_M4_omega04_230))/(np.pi*(omega04_230**2)))**2)
y5_230 = omega05_230*np.sqrt(1+((lambda230*(x5_230-z_sub_omega01_230-z_omega01_230_M2-z_M2_omega02_230-
    z_omega02_230_M3-z_M3_omega03_230-z_omega03_230_M4-z_M4_omega04_230-z_omega04_230_M6-z_M6_omega05_230))/
    (np.pi*(omega05_230**2)))**2)

y1_345 = omega01_345*np.sqrt(1+((lambda345*(x1_345-z_sub_omega01_345))/(np.pi*(omega01_345**2)))**2)
y2_345 = omega02_345*np.sqrt(1+((lambda345*(x2_345-z_sub_omega01_345-z_omega01_345_M2-z_M2_omega02_345))/
    (np.pi*(omega02_345**2)))**2)
y3_345 = omega03_345*np.sqrt(1+((lambda345*(x3_345-z_sub_omega01_345-z_omega01_345_M2-z_M2_omega02_345-
    z_omega02_345_M3-z_M3_omega03_345))/(np.pi*(omega03_345**2)))**2)
y4_345 = omega04_345*np.sqrt(1+((lambda345*(x4_345-z_sub_omega01_345-z_omega01_345_M2-z_M2_omega02_345-
    z_omega02_345_M3-z_M3_omega03_345-z_omega03_345_M4-z_M4_omega04_345))/(np.pi*(omega04_345**2)))**2)
y5_345 = omega05_345*np.sqrt(1+((lambda345*(x5_345-z_sub_omega01_345-z_omega01_345_M2-z_M2_omega02_345-
    z_omega02_345_M3-z_M3_omega03_345-z_omega03_345_M4-z_M4_omega04_345-z_omega04_345_M6-z_M6_omega05_345))/
    (np.pi*(omega05_345**2)))**2)

x6 = l_sub_M2+l_M2_M3+l_M3_M4+130.388 # 副鏡からDewar窓までの距離
```
